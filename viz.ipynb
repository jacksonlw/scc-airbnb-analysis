# Visualizations of Santa Clara County Airbnbs

**IMPORTANT**: Must download the `scc_airbnbs.csv` using the `download_scc_airbnbs.ipynb` before running this notebook


```python
import pandas as pd
import altair as alt
import json
```

```python
df = pd.read_csv('scc_airbnbs.csv')
df = df.set_index('id')

# Convert stringified JSON columns to Python objects
df['amenities'] = df['amenities'].apply(json.loads)
df['host_verifications'] = df['host_verifications'].apply(json.loads)

# Convert f and t to False and True
df = df.replace({'f': False, 't': True})
```

```python
df.columns
```

```python
price_vs_ammenities = alt.Chart(df).mark_point().transform_calculate(
    amenities_count="datum.amenities.length"
).encode(
    x='amenities_count:Q',
    y=alt.Y('price:Q').scale(type='log'),
    color='room_type:N',
    tooltip=['name:N', 'amenities_count:Q', 'price:Q']
).properties(
    width=600,
    height=400,
    title='Price vs Amenities Count'
)

price_vs_ammenities
```

```python
host_since_vs_reviews = alt.Chart(df).mark_bar().transform_timeunit(
    field='host_since',
    timeUnit='year',
    as_='host_since_year'
).transform_aggregate(
    avg_review="mean(review_scores_rating)",
    groupby=["host_since_year"]
).encode(
    x=alt.X('host_since_year:T', axis=alt.Axis(format='%Y'), title='Host Since Year'),
    y=alt.Y('avg_review:Q', title='Average Review Score', scale=alt.Scale(domain=[4.5, 5], clamp=True)),
).properties(
    width=600,
    height=400,
    title='Host Since vs Average Review Score'
)

host_since_vs_reviews
```
